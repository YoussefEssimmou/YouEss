```python
# Importation 
import os
import random
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from torchvision.utils import make_grid
from PIL import Image
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

# Configuration
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
PRINT_EVERY = 1

# Paramètres 
DATA_DIR = "Downloads/archive/AMHCD_64/AMHCD_64"  
BATCH_SIZE = 64
IMG_SIZE = 32   
NUM_CLASSES = 33
NUM_EPOCHS = 30
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4
USE_BATCHNORM = True
USE_DROPOUT = False
CHECKPOINT_DIR = "./checkpoints"
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

# Dataset 
class AMHCDDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.samples = []
        self.labels_map = {}
        classes = sorted([d for d in os.listdir(root_dir) if os.path.isdir(os.path.join(root_dir, d))])
        for idx, cl in enumerate(classes):
            self.labels_map[idx] = cl
            cl_dir = os.path.join(root_dir, cl)
            for fname in os.listdir(cl_dir):
                if fname.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp')):
                    self.samples.append((os.path.join(cl_dir, fname), idx))
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        path, label = self.samples[idx]
        img = Image.open(path).convert("RGB")
        if self.transform:
            img = self.transform(img)
        return img, label

# Transforms (train/val/test)
train_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5,0.5,0.5], std=[0.5,0.5,0.5]),
])

val_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5,0.5,0.5], std=[0.5,0.5,0.5]),
])

# Split dataset to (train/val/test) 
def make_splits(data_dir, val_ratio=0.15, test_ratio=0.15):
    # Build full dataset then split by indices
    full_ds = AMHCDDataset(data_dir, transform=None)
    n = len(full_ds)
    indices = list(range(n))
    random.shuffle(indices)
    n_test = int(n * test_ratio)
    n_val = int(n * val_ratio)
    test_idx = indices[:n_test]
    val_idx = indices[n_test:n_test+n_val]
    train_idx = indices[n_test+n_val:]
    # Create Subsets that apply transforms on the fly
    from torch.utils.data import Subset
    train_ds = Subset(AMHCDDataset(data_dir, transform=train_transform), train_idx)
    val_ds = Subset(AMHCDDataset(data_dir, transform=val_transform), val_idx)
    test_ds = Subset(AMHCDDataset(data_dir, transform=val_transform), test_idx)
    # Extract label->name mapping (from full dataset)
    labels_map = full_ds.labels_map
    return train_ds, val_ds, test_ds, labels_map

# LeNet-5 adapté 
class LeNet5_RGB(nn.Module):
    def __init__(self, num_classes=NUM_CLASSES, use_bn=USE_BATCHNORM, dropout=USE_DROPOUT):
        super().__init__()
        # Conv layers
        self.conv1 = nn.Conv2d(3, 6, kernel_size=5, stride=1, padding=0)   # => (32-5+1)=28
        self.conv2 = nn.Conv2d(6, 16, kernel_size=5)                      # => (14-5+1)=10 after pool
        # optional BatchNorm
        self.bn1 = nn.BatchNorm2d(6) if use_bn else nn.Identity()
        self.bn2 = nn.BatchNorm2d(16) if use_bn else nn.Identity()
        # Fully connected
        # after conv + pools: feature map size calculation:
        # conv1 -> relu -> avgpool(2) => 28 -> 14
        # conv2 -> relu -> avgpool(2) => 10 -> 5
        # maps: 16 channels * 5 * 5
        self.fc1 = nn.Linear(16 * 5 * 5, 120)
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, num_classes)
        self.pool = nn.AvgPool2d(kernel_size=2, stride=2)
        self.dropout = nn.Dropout(0.5) if dropout else nn.Identity()

    def forward(self, x, return_feature_maps=False):
        x = F.relu(self.bn1(self.conv1(x)))
        feat1 = x
        x = self.pool(x)
        x = F.relu(self.bn2(self.conv2(x)))
        feat2 = x
        x = self.pool(x)
        x = x.view(x.size(0), -1)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        if return_feature_maps:
            return x, [feat1, feat2]
        return x

# Training and evaluation utilities
def save_checkpoint(state, filename):
    torch.save(state, filename)

def load_checkpoint(path, model, optimizer=None):
    ckpt = torch.load(path, map_location=DEVICE)
    model.load_state_dict(ckpt['model_state'])
    if optimizer and 'opt_state' in ckpt:
        optimizer.load_state_dict(ckpt['opt_state'])
    return ckpt

def train_one_epoch(model, optimizer, criterion, dataloader, device):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    for imgs, labels in tqdm(dataloader, desc="Train", leave=False):
        imgs = imgs.to(device)
        labels = labels.to(device)
        optimizer.zero_grad()
        outputs = model(imgs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * imgs.size(0)
        preds = outputs.argmax(dim=1)
        correct += (preds == labels).sum().item()
        total += imgs.size(0)
    epoch_loss = running_loss / total
    epoch_acc = correct / total
    return epoch_loss, epoch_acc

def eval_one_epoch(model, criterion, dataloader, device):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0
    all_preds = []
    all_labels = []
    with torch.no_grad():
        for imgs, labels in tqdm(dataloader, desc="Eval", leave=False):
            imgs = imgs.to(device)
            labels = labels.to(device)
            outputs = model(imgs)
            loss = criterion(outputs, labels)
            running_loss += loss.item() * imgs.size(0)
            preds = outputs.argmax(dim=1)
            correct += (preds == labels).sum().item()
            total += imgs.size(0)
            all_preds.append(preds.cpu().numpy())
            all_labels.append(labels.cpu().numpy())
    epoch_loss = running_loss / total
    epoch_acc = correct / total
    all_preds = np.concatenate(all_preds)
    all_labels = np.concatenate(all_labels)
    return epoch_loss, epoch_acc, all_preds, all_labels

# Full training loop
def fit(model, train_loader, val_loader, epochs=NUM_EPOCHS, lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY, scheduler=None, ckpt_path=None):
    model = model.to(DEVICE)
    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    history = {"train_loss":[], "train_acc":[], "val_loss":[], "val_acc":[]}
    best_val_acc = 0.0

    for epoch in range(1, epochs+1):
        train_loss, train_acc = train_one_epoch(model, optimizer, criterion, train_loader, DEVICE)
        val_loss, val_acc, _, _ = eval_one_epoch(model, criterion, val_loader, DEVICE)

        history["train_loss"].append(train_loss)
        history["train_acc"].append(train_acc)
        history["val_loss"].append(val_loss)
        history["val_acc"].append(val_acc)

        if scheduler is not None:
            scheduler.step(val_loss)

        if (epoch % PRINT_EVERY == 0) or (epoch == epochs):
            print(f"Epoch {epoch}/{epochs}  Train loss: {train_loss:.4f} acc: {train_acc:.4f}  |  Val loss: {val_loss:.4f} acc: {val_acc:.4f}")

        # Save best
        if ckpt_path:
            if val_acc > best_val_acc:
                best_val_acc = val_acc
                save_checkpoint({
                    'epoch': epoch,
                    'model_state': model.state_dict(),
                    'opt_state': optimizer.state_dict(),
                    'val_acc': val_acc,
                    'history': history
                }, ckpt_path)
    return history

# Plot utilities
def plot_history(history):
    epochs = len(history["train_loss"])
    plt.figure(figsize=(12,4))
    plt.subplot(1,2,1)
    plt.plot(range(1, epochs+1), history["train_loss"], label="train")
    plt.plot(range(1, epochs+1), history["val_loss"], label="val")
    plt.title("Loss")
    plt.legend()
    plt.subplot(1,2,2)
    plt.plot(range(1, epochs+1), history["train_acc"], label="train")
    plt.plot(range(1, epochs+1), history["val_acc"], label="val")
    plt.title("Accuracy")
    plt.legend()
    plt.show()

def plot_confusion(y_true, y_pred, labels_map, normalize=True, figsize=(12,10)):
    cm = confusion_matrix(y_true, y_pred)
    if normalize:
        cm = cm.astype('float') / (cm.sum(axis=1)[:, np.newaxis] + 1e-12)
    plt.figure(figsize=figsize)
    sns.heatmap(cm, cmap='viridis', xticklabels=[labels_map[i] for i in sorted(labels_map)], yticklabels=[labels_map[i] for i in sorted(labels_map)], vmin=0, vmax=1 if normalize else None)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title("Confusion matrix")
    plt.show()

# Feature map visualization helper
def visualize_feature_maps(model, img_tensor, device=DEVICE, max_maps=6):
    model.eval()
    x = img_tensor.unsqueeze(0).to(device)
    with torch.no_grad():
        out, feats = model(x, return_feature_maps=True)
    # feats is a list [feat1, feat2]
    for i, f in enumerate(feats):
        f = f.cpu().squeeze(0)  # (C, H, W)
        n_maps = min(f.shape[0], max_maps)
        fig, axs = plt.subplots(1, n_maps, figsize=(n_maps*2,2))
        for j in range(n_maps):
            ax = axs[j] if n_maps>1 else axs
            fmap = f[j].numpy()
            ax.imshow((fmap - fmap.min())/(fmap.max()-fmap.min()+1e-8), cmap='gray')
            ax.axis('off')
        plt.suptitle(f"Feature maps - layer {i+1}")
        plt.show()

# Test
if __name__ == "__main__":
    # 1) Prepare dataset (modify DATA_DIR to your local path)
    if not os.path.isdir(DATA_DIR):
        raise ValueError(f"DATA_DIR '{DATA_DIR}' not found. Téléchargez et extrayez AMHCD puis mettez le chemin ici.")
    train_ds, val_ds, test_ds, labels_map = make_splits(DATA_DIR, val_ratio=0.15, test_ratio=0.15)
    print(f"Train: {len(train_ds)}, Val: {len(val_ds)}, Test: {len(test_ds)}")

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    # 2) Model
    model = LeNet5_RGB(num_classes=len(labels_map), use_bn=USE_BATCHNORM, dropout=USE_DROPOUT)
    print(model)

    # 3) Train
    ckpt_path = os.path.join(CHECKPOINT_DIR, "best_lenet5_amhcd.pth")
    history = fit(model, train_loader, val_loader, epochs=NUM_EPOCHS, lr=LEARNING_RATE, ckpt_path=ckpt_path)

    # 4) Plot history
    plot_history(history)

    # 5) Load best model and evaluate on test
    load_checkpoint(ckpt_path, model)
    criterion = nn.CrossEntropyLoss()
    test_loss, test_acc, test_preds, test_labels = eval_one_epoch(model, criterion, test_loader, DEVICE)
    print(f"Test loss: {test_loss:.4f}  Test acc: {test_acc:.4f}")

    # 6) Classification report (per-class precision/recall/f1)
    print(classification_report(test_labels, test_preds, target_names=[labels_map[i] for i in sorted(labels_map)], zero_division=0))

    # 7) Confusion matrix (normalisée)
    plot_confusion(test_labels, test_preds, labels_map, normalize=True)

    # 8) Visualiser quelques images & feature maps
    # Récupérer un batch
    imgs_batch, labels_batch = next(iter(test_loader))
    # Montrer quelques images
    def imshow_batch(imgs, labels, labels_map, n=8):
        imgs = imgs[:n]
        labels = labels[:n]
        # Denormalize
        imgs = imgs * 0.5 + 0.5
        grid = make_grid(imgs, nrow=n)
        plt.figure(figsize=(12,3))
        plt.imshow(np.transpose(grid.numpy(), (1,2,0)))
        plt.axis('off')
        plt.show()
        print("Labels:", [labels_map[int(l.item())] for l in labels])

    imshow_batch(imgs_batch, labels_batch, labels_map, n=8)

    # Visualize feature maps for first image
    img0 = imgs_batch[0]  # already normalized
    visualize_feature_maps(model, img0)

    # 9) Save final model (optional)
    final_path = os.path.join(CHECKPOINT_DIR, "final_lenet5_amhcd.pth")
    save_checkpoint({
        'model_state': model.state_dict(),
        'labels_map': labels_map
    }, final_path)
    print("Saved final model to", final_path)
```
